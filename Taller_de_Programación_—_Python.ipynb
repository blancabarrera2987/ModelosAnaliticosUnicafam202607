```python
# Calculadora de descuento Una tienda aplica descuentos según el monto: 0% si es menos de $50,000; 10% entre $50,000 y $99,999; 20% entre $100,000 y $199,999; 30% si es $200,000 o más. Dado monto = 120000, calcula e imprime el descuento y el total a pagar.

monto = 120000

if monto < 50000:
    descuento = 0
elif monto < 100000:
    descuento = 0.10
elif monto < 200000:
    descuento = 0.20
else:
    descuento = 0.30

ahorro = monto * descuento
total  = monto - ahorro
print(f"Descuento: {int(descuento*100)}% (${ahorro:,.0f})")
print(f"Total: ${total:,.0f}")
# Descuento: 20% ($24,000) — Total: $96,000
```

```python
# FizzBuzz clásico Para un número n, imprime: "FizzBuzz" si es múltiplo de 3 y de 5; "Fizz" si es múltiplo solo de 3; "Buzz" si es múltiplo solo de 5; el número si no cumple ninguna. Prueba con n = 15, n = 9, n = 10 y n = 7.


def fizzbuzz(n):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)

fizzbuzz(15)  # FizzBuzz
fizzbuzz(9)   # Fizz
fizzbuzz(10)  # Buzz
fizzbuzz(7)   # 7
```

```python
# Suma acumulada Usando un ciclo for, suma todos los números del 1 al 100 e imprime el resultado. Luego suma solo los números pares del mismo rango.

# Suma 1 al 100
total = 0
for i in range(1, 101):
    total += i
print(total)   # 5050

# Solo pares — opción 1: range con paso 2
pares = 0
for i in range(2, 101, 2):
    pares += i
print(pares)   # 2550
```

```python
# Recorrer una lista Dada precios = [12000, 45000, 8500, 67000, 3200], usa un ciclo para: (1) imprimir cada precio con formato Item 1: $12,000, (2) imprimir al final cuántos artículos cuestan más de $10,000.

precios = [12000, 45000, 8500, 67000, 3200]
caros = 0

for i, p in enumerate(precios, 1):
    print(f"Item {i}: ${p:,}")
    if p > 10000:
        caros += 1

print(f"Artículos > $10,000: {caros}")  # 3
```

```python
# While con condición Un banco tiene saldo = 1000. Cada mes retiras 150 y se cobran 20 de comisión. Usa un ciclo while para simular cuántos meses pasan antes de que el saldo llegue a cero o menos. Imprime el saldo al final de cada mes.

saldo = 1000
mes   = 0

while saldo > 0:
    saldo -= 150   # retiro
    saldo -= 20    # comisión
    mes   += 1
    print(f"Mes {mes}: saldo ${saldo}")

print(f"Saldo agotado en {mes} meses")
# Mes 1: 830, Mes 2: 660 ... Mes 6: 30, Mes 7: -140
```

```python
# break y continue Dada la lista numeros = [3, 7, -2, 15, 4, -8, 9, 1], escribe un ciclo que: (1) saltee los números negativos con continue, (2) se detenga con break al encontrar un número mayor a 10, (3) imprima los números que procesa.

numeros = [3, 7, -2, 15, 4, -8, 9, 1]

for n in numeros:
    if n < 0:
        continue         # salta negativos
    if n > 10:
        print(f"Encontré {n}, deteniendo.")
        break            # para el ciclo
    print(f"Procesando: {n}")

# Procesando: 3
# Procesando: 7
# (-2 saltado)
# Encontré 15, deteniendo.
```
