Crea un programa que verifique si una persona es mayor de edad. Una persona es considerada mayor de edad si tiene 18 años o más.

```python
print("------VERIFICACION DE DATOS-----")
edad=int(input("Ingresa tu Edad"))

if edad<18:
  print(f"eres menor de edad, tienes:{edad}")
else:
  print(f"eres mayor de edad,tienes:{edad}")
```

Escribe un programa que verifique si un número ingresado por el usuario es par o impar.

```python
# Ingrese un número entero

entrada = input("Por favor, ingresa un número entero: ")
numero = int(entrada)
if numero % 2 == 0:
  print(f"El número {numero} es PAR.")
else:
 print(f"El número {numero} es IMPAR.")
```

Crea un programa que lea dos números y determine cuál es mayor, cuál es menor o si ambos son iguales.

```python
entrada1 = input("Ingresa el primer número: ")
entrada2 = input("Ingresa el segundo número: ")

numero1 = float(entrada1)
numero2 = float(entrada2)

if numero1 > numero2:
  print(f"\nEl primer número ({numero1}) es MAYOR que el segundo número ({numero2}).")
  print(f"El menor es: {numero2}")
elif numero1 < numero2:
  print(f"\nEl segundo número ({numero2}) es MAYOR que el primer número ({numero1}).")
  print(f"El menor es: {numero1}")
else:
  print(f"\nAmbos números son IGUALES ({numero1} = {numero2}).")
```

Crea un programa que determine si un número ingresado por el usuario está en el rango de 1 a 100 (inclusive) y si es divisible por 5.

```python
try:
    numero = int(input("Ingresa un número entero: "))
    if 1 <= numero <= 100:
        if numero % 5 == 0:
            print(f"El número {numero} está en el rango (1-100) y es divisible por 5.")
        else:
            print(f"El número {numero} está en el rango (1-100), pero no es divisible por 5.")
    else:
        print(f"El número {numero} está fuera del rango permitido (1-100).")
except ValueError:
    print("Error: Por favor, ingresa un número entero válido.")
```
