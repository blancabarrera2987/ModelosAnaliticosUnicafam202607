**CREACION Y MODIFICACION DE LISTAS**

```python
# 1. Crea una lista llamada números que contenga los números del 1 al 5.
mis_listados=[]
# lista con varios elementos
mis_listados=[1,2,3,4,5,]
print(mis_listados)
```

```python
# 2. Añade el número 6 al final de la lista numeros.
mis_listados.append("6")
print(mis_listados)
```

```python
# 3. Cambia el segundo elemento de la lista numeros a 10.

mis_listados[2]="10"
print(mis_listados)
```

```python
# 4. Elimina el cuarto elemento de la lista números.

mis_listados.remove("6")
print(mis_listados)
```
