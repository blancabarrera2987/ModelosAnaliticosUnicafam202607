```python
# Datos base
datos_personales = {
    "Nombre": "Blanca Barrera Florez",
    "ciudad de Nacimiento": "Bucaramanga",
    "Cuidad de Residencia": "Bogota",
    "Profesion": "Administracion de Empresas",
    "Direccion de Residencia": "Calle 30C sur 11A 25",
    "Empresa Donde Labora": "Aguas de Bogota"
}

# 1. Agregar un dato nuevo
clave = input("Ingresa la nueva clave: ")
valor = input("Ingresa el valor para la clave: ")
datos_personales[clave] = valor
print("Diccionario actualizado:", datos_personales)
```

```python
# 2. Actualizar un dato existente

datos_personales = {
    "Nombre": "Blanca Barrera Florez",
    "ciudad de Nacimiento": "Bucaramanga",
    "Cuidad de Residencia": "Bogota",
    "Profesion": "Administracion de Empresas",
    "Direccion de Residencia": "Calle 30C sur 11A 25",
    "Empresa Donde Labora": "Aguas de Bogota"
}

clave_a_actualizar = input("Ingresa la clave que deseas actualizar: ")

if clave_a_actualizar in datos_personales:
    nuevo_valor = input(f"La clave '{clave_a_actualizar}' existe. Ingresa el nuevo valor: ")
    datos_personales[clave_a_actualizar] = nuevo_valor
    print("¡Actualización exitosa!")
    print("Diccionario actualizado:", datos_personales)
else:
    print(f"Error: La clave '{clave_a_actualizar}' no se encuentra en el diccionario.")
```

```python
# 3. Eliminar un dato con del

datos_personales = {
    "Nombre": "Blanca Barrera Florez",
    "ciudad de Nacimiento": "Bucaramanga",
    "Cuidad de Residencia": "Bogota",
    "Profesion": "Administracion de Empresas",
    "Direccion de Residencia": "Calle 30C sur 11A 25",
    "Empresa Donde Labora": "Aguas de Bogota"
}


clave_a_eliminar = input("Ingresa la clave que deseas eliminar: ")

if clave_a_eliminar in datos_personales:
    del datos_personales[clave_a_eliminar]
    print(f"La clave '{clave_a_eliminar}' ha sido eliminada.")
    print("Diccionario actualizado:", datos_personales)
else:
    print(f"Error: La clave '{clave_a_eliminar}' no existe.")
```

```python
# 4. Eliminar un dato con .pop()

clave_a_eliminar = input("Ingresa la clave que deseas eliminar con .pop(): ")

if clave_a_eliminar in datos_personales:
    valor_eliminado = datos_personales.pop(clave_a_eliminar)
    print(f"La clave '{clave_a_eliminar}' fue eliminada y su valor era: {valor_eliminado}")
    print("Diccionario actualizado:", datos_personales)
else:
    print(f"Error: La clave '{clave_a_eliminar}' no existe.")

```

```python
# 5. Consultar un dato con .get()

clave_a_consultar = input("Ingresa la clave que deseas consultar: ")

resultado = datos_personales.get(clave_a_consultar, "No disponible")
print(f"Resultado de la consulta: {resultado}")
```

```python
# 6. Ver el perfil completo
print("Perfil completo:", datos_personales)
```

```python


print("--- MENÚ DE OPCIONES ---")
print("1. Agregar un dato nuevo")
print("2. Actualizar un dato existente")
print("3. Eliminar un dato con del")
print("4. Eliminar un dato con .pop()")
print("5. Consultar un dato con .get()")
print("6. Ver el perfil completo")

opcion = input("Elige una opción (1-6): ")

if opcion == "1":
    clave = input("Ingresa la nueva clave: ")
    valor = input("Ingresa el valor para la clave: ")
    datos_personales[clave] = valor
    print("Diccionario actualizado:", datos_personales)
elif opcion == "2":
    clave = input("Ingresa la clave a actualizar: ")
    if clave in datos_personales:
        nuevo_valor = input(f"Ingresa el nuevo valor para {clave}: ")
        datos_personales[clave] = nuevo_valor
        print("Valor actualizado.")
    else:
        print("La clave no existe.")
elif opcion == "3":
    clave = input("Ingresa la clave a eliminar con 'del': ")
    if clave in datos_personales:
        del datos_personales[clave]
        print("Clave eliminada.")
    else:
        print("La clave no existe.")
elif opcion == "4":
    clave = input("Ingresa la clave a eliminar con '.pop()': ")
    if clave in datos_personales:
        valor = datos_personales.pop(clave)
        print(f"Eliminado: {valor}")
    else:
        print("La clave no existe.")
elif opcion == "5":
    clave = input("Ingresa la clave a consultar: ")
    print("Resultado:", datos_personales.get(clave, "No disponible"))
elif opcion == "6":
    print("Perfil completo:", datos_personales)
else:
    print("Opción inválida.")
```
