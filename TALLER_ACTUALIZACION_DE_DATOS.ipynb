TALLER ACTUALIZACION DE DATOS

```python
Estudiante={
    "Nombres":"Blanca Barrera Florez",
    "ciudad de Nacimiento":"Bucaramanga",
    "Cuidad de Residencia":"Bogota",
    "Profesion":"Administracion de Empresas",
    "Direccion de Residencia":"Calle 30C sur 11A 25",
    "Empresa Donde Labora":"Aguas de Bogota"
    }
```

```python
# CONSTRUCCION DEL MENU

print("----Bienvenido a tu programa de Actualizacion de Datos")

print("Para Comenzar Elije la Opcion Deseada")

print("1. Agregar un Dato Nuevo")
print("2. Actualizar un Datos")
print("3. Eliminar un Dato con Del")
print("4. Eliminar un Dato con Pop")
print("5. Consultar un Dato")
print("6. Imprimiar Diccionario Completo")
seleccion=int(input("Selecciona la Opccion Deseada:"))
```

```python
# AGREGAR UN DATO NUEVO

if seleccion==1:
  clave=input("Ingrese la Clave Deseada:")
  valor=input("Ingrese el Valor Deseado:")

  # AGREGAR UN ELEMENTO NUEVO

  Estudiante[clave]=valor
  print("el valor se agrego exitosamente")
  print(Estudiante)

  # ACTUALIZAR UN ELEMENTO NUEVO

elif seleccion==2:
  clave=input("Ingrese la Clave a Actualizar")
  Estudiante[clave]=input("Ingrese Valor Nuevo")
  print("el Fue Actualizado con Exito")
  print(Estudiante)

 # ELIMINAR UN ELEMENTO NUEVO (DEL)

elif seleccion==3:
  clave=input("Ingrese la Clave a Eliminar")
  del Estudiante [clave]
  print("El Dato fue Eliminado Exitosamente")

# ELIMINAR UN ELEMENTO NUEVO (POP)

elif seleccion==4:
 clave=input("Ingrese la Clave a Eliminar")
 Estudiante.pop(clave)
 print("El Dato fue Eliminado Exitosamente")
 print(Estudiante)

# CONSULTAR DATO

elif seleccion==5:
 clave=input("Ingrese la Clave a Consultar")
 valor=Estudiante.get(clave, "no disponible")
 print(f"{clave}:{valor}")

# IMPRIMIR DICCIONARIO

elif seleccion==6:
  print("Perfil Completo")
  print(Estudiante)

else:
  print("Opccion Invalida")
```
